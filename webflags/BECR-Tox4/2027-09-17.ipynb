```python
import pandas as pd
```

```python
cluster_df = pd.read_csv('results_flankgene.fasta_cluster_out_5_1e-05_outdesc.txt', sep='\t', header=None)
operon_df = pd.read_csv('results_TreeOrder_operon.tsv', sep='\t', header=None)
```

```python
cluster_df.head()
```

```python
cluster_headers = ['cluster_num', 'pid', 'description']
cluster_df.columns = cluster_headers
cluster_df.head()
```

```python
operon_df.head()
```

```python
operon_header = ['Query', 'Length (aa)', 'Query strand', 
                 'Direction relative to query', 'Cluster Number', 
                 'Start position relative to query', 
                 'End position relative to query', 'Start position in genome',
                 'End position in genome',
                 'Protein accessions and order for output',
                 'Genome accesssion', 'Genome Assembly Identifier']

operon_df.columns = operon_header
operon_df.head(20)
```

```python
cluster_df.head()
```

```python
# add description column to main table
pid = [x.split('#')[0] for x in operon_df['Protein accessions and order for output']]
```

```python
operon_df['pid'] = pid
```

```python
just_desc = [' '.join(x.split(' ')[1:]).rstrip() for x in cluster_df['description']]
```

```python
cluster_df['just_desc'] = just_desc
```

```python
cluster_df.head()
```

```python
operon_df = pd.merge(operon_df, cluster_df[['pid', 'just_desc']], how='left', on='pid')
```

```python
operon_df = operon_df.drop(columns=['pid']).rename(columns=({'just_desc':'Protein Description'}))
```

```python
operon_df.to_csv('BECR-Tox4_gene_neighborhood.csv', index=None)
```
